```python
import heapq

def dijkstra(graph, start):
    n = len(graph)
    dist = [float('inf')] * n
    dist[start] = 0
    pq = [(0, start)]
    vis = [False] * n

    while pq:
        d, u = heapq.heappop(pq)
        if vis[u]: continue
        vis[u] = True
        print(f"Visit {u} with dist {d}")

        for v, w in graph[u]:
            if dist[v] > d + w:
                dist[v] = d + w
                heapq.heappush(pq, (dist[v], v))
    return dist

# Adjacency list: node -> [(neighbor, weight)]
graph = [
    [(1, 4), (2, 1)],
    [(3, 1)],
    [(1, 2), (3, 5)],
    []
]
print(dijkstra(graph, 0)) # [0, 3, 1, 4]
```
