```python
class SegTree:
    def __init__(self, sz):
        self.n = 1
        while self.n < sz:
            self.n <<= 1
        self.t = [0] * (2 * self.n)

    def build(self, a):
        for i in range(len(a)):
            self.t[self.n + i] = a[i]
        for i in range(self.n - 1, 0, -1):
            self.t[i] = self.t[i << 1] + self.t[i << 1 | 1]

    def update(self, p, v):
        p += self.n
        self.t[p] = v
        p >>= 1
        while p > 0:
            self.t[p] = self.t[p << 1] + self.t[p << 1 | 1]
            p >>= 1

    def query(self, l, r): # [l, r] inclusive
        res = 0
        l += self.n
        r += self.n + 1
        while l < r:
            if l & 1:
                res += self.t[l]
                l += 1
            if r & 1:
                r -= 1
                res += self.t[r]
            l >>= 1
            r >>= 1
        return res

if __name__ == "__main__":
    a = [1, 3, 5, 7, 9, 11]
    st = SegTree(6)
    st.build(a)
    print(st.query(1, 3)) # 3+5+7 = 15
    st.update(1, 10) # a[1] = 10
    print(st.query(1, 3)) # 10+5+7 = 22
```
